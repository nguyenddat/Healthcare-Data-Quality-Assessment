```python
def profile_table(df, name, cat_threshold=30):
    rows = []
    for col in df.columns:
        series = df[col]
        dtype = series.dtype
        n_unique = series.nunique(dropna=True)
        n_null = series.isna().sum()

        row = {
            'table': name,
            'column': col,
            'dtype': dtype,
            'null_count': n_null,
            'unique_count': n_unique,
            'type_category': None,
            'min': None,
            'max': None,
            'mean': None,
            'sample_values': None
        }

        if pd.api.types.is_numeric_dtype(series):
            row['type_category'] = 'NUM'
            row['min'] = series.min()
            row['max'] = series.max()
            row['mean'] = round(series.mean(), 2) if n_unique > 0 else None
        else:
            if n_unique <= cat_threshold:
                row['type_category'] = 'CAT'
                row['sample_values'] = series.dropna().unique().tolist()
            else:
                row['type_category'] = 'TEXT/HIGH-CARD'

        rows.append(row)

    return pd.DataFrame(rows)
```

```python
def detail_values_table(df, name, cat_threshold=30, high_card_only=False):
    rows = []
    for col in df.columns:
        series = df[col]
        if pd.api.types.is_numeric_dtype(series):
            continue
        n_unique = series.nunique(dropna=True)
        is_high_card = n_unique > cat_threshold
        if high_card_only and not is_high_card:
            continue
        values = ", ".join(map(str, series.dropna().unique()))        
        rows.append({
            'table': name,
            'column': col,
            'type_category': 'TEXT/HIGH-CARD' if is_high_card else 'CAT',
            'value': values
        })
    return pd.DataFrame(rows)
```

```python
value_results = []

value_results.append(detail_values_table(tb_chamber, "tb_chamber"))
value_results.append(detail_values_table(tb_department, "tb_department"))
value_results.append(detail_values_table(tb_dm_benhvien, "tb_dm_benhvien"))
value_results.append(detail_values_table(tb_dm_icd10, "tb_dm_icd10"))
value_results.append(detail_values_table(tb_dm_servicesubgroup, "tb_dm_servicesubgroup"))
value_results.append(detail_values_table(tb_mediboxdocument, "tb_mediboxdocument"))
value_results.append(detail_values_table(tb_mediboxdocument_sign, "tb_mediboxdocument_sign"))
value_results.append(detail_values_table(tb_medicalrecord_nhandinhbandau, "tb_medicalrecord_nhandinhbandau"))
value_results.append(detail_values_table(tb_medicalrecord_ylenh, "tb_medicalrecord_ylenh"))
value_results.append(detail_values_table(tb_nhanvien, "tb_nhanvien"))
value_results.append(detail_values_table(tb_nhanvienactivity, "tb_nhanvienactivity"))
value_results.append(detail_values_table(tb_patientrecord_chuyendt, "tb_patientrecord_chuyendt"))
value_results.append(detail_values_table(tb_room, "tb_room"))
value_results.append(detail_values_table(tb_service, "tb_service"))
value_results.append(detail_values_table(tb_treatment, "tb_treatment"))
value_results.append(detail_values_table(tb_servicedata_pttt, "tb_servicedata_pttt"))
value_results.append(detail_values_table(tb_medicalrecord, "tb_medicalrecord"))
value_results.append(detail_values_table(tb_medicalrecord_hoichan_phauth, "tb_medicalrecord_hoichan_phauth"))
value_results.append(detail_values_table(tb_medicalrecord_khambenh, "tb_medicalrecord_khambenh"))

final_profile_values = pd.concat(value_results, ignore_index=True)
final_profile_values['value'] = final_profile_values['value'].astype(str)
final_profile_values.to_csv("profile_values.csv", index=False, encoding='utf-8-sig')
```
